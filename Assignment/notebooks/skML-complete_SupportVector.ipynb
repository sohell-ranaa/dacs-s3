* _Quick visual check of unique values, deal with unique identifiers_

```python
# Identify columns with only one value 
# or with number of unique values == number of rows
n_eq_one = []
n_eq_all = []

print('Unique value count: Train (',train_df.shape[0],'rows ) ~ Test(',test_df.shape[0],'rows )')
for col in train_df.columns:
    lctrn = len(train_df[col].unique())
    lctst = len(test_df[col].unique())
    print(col, ' ::> ', lctrn, ' ~ ', lctst)
    if (lctrn == 1) and (lctrn == lctst): 
        n_eq_one.append(train_df[col].name)
    if lctrn == train_df.shape[0]:
        n_eq_all.append(train_df[col].name)
```

```python
# Drop columns with only one value
if len(n_eq_one) > 0:
    print('Dropping single-valued features')
    print(n_eq_one)
    train_df.drop(n_eq_one, axis=1, inplace=True)
    test_df.drop(n_eq_one, axis=1, inplace=True)

# Drop or bin columns with number of unique values == number of rows
if len(n_eq_all) > 0:
    print('Dropping unique identifiers')
    print(n_eq_all)
    train_df.drop(n_eq_all, axis=1, inplace=True)
    test_df.drop(n_eq_all, axis=1, inplace=True)

# continue with featue selection / feature engineering
```

## IMPORTANT: Optimized SVM Approach

**Challenge with SVM**: SVC is very slow with large datasets (O(n^2) to O(n^3))

**Solutions**:
1. Use LinearSVC for faster training (linear kernel only)
2. Apply class_weight='balanced' for imbalanced data
3. Tune regularization parameter C

```python
# ============================================================
# IMPROVED HYPERPARAMETER TUNING - SUPPORT VECTOR CLASSIFIERS
# ============================================================

from sklearn.svm import SVC, LinearSVC
from sklearn.linear_model import RidgeClassifier
from sklearn.metrics import balanced_accuracy_score, f1_score, accuracy_score, matthews_corrcoef

print("="*70)
print("COMPREHENSIVE SUPPORT VECTOR CLASSIFIER COMPARISON")
print("="*70)
print("Note: SVC can be slow. LinearSVC and Ridge are much faster.")

all_results = {}
```

```python
# ============================================================
# STEP 1: Baseline Comparison
# ============================================================

# For SVC, we use a sample to speed up (SVC is O(n^2) to O(n^3))
from sklearn.model_selection import train_test_split
X_train_small, _, y_train_small, _ = train_test_split(
    X_train, y_train, train_size=15000, random_state=42, stratify=y_train
)
print(f"Using {len(X_train_small)} samples for SVC (faster training)")

baseline_models = [
    ("SVC_baseline", SVC(random_state=42), True),  # True = use small sample
    ("LinearSVC_baseline", LinearSVC(random_state=42, max_iter=2000), False),
    ("Ridge_baseline", RidgeClassifier(random_state=42), False),
]

for name, clf, use_small in baseline_models:
    print(f"\n>>> {name}")
    trs = time()
    if use_small:
        clf.fit(X_train_small, y_train_small)
    else:
        clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    tre = time() - trs
    
    all_results[name] = {
        'accuracy': accuracy_score(y_test, y_pred),
        'balanced_accuracy': balanced_accuracy_score(y_test, y_pred),
        'f1_score': f1_score(y_test, y_pred, average='weighted'),
        'mcc': matthews_corrcoef(y_test, y_pred),
        'time': tre,
        'predictions': y_pred
    }
    print(f"    Time: {tre:.1f}s | Acc: {all_results[name]['accuracy']:.4f} | Bal_Acc: {all_results[name]['balanced_accuracy']:.4f}")
```

```python
# ============================================================
# STEP 2: Tuned SVC (with class balancing)
# ============================================================

tuned_svc = [
    ("SVC_balanced", SVC(class_weight='balanced', random_state=42), True),
    ("SVC_C10", SVC(C=10, random_state=42), True),
    ("SVC_rbf_balanced", SVC(kernel='rbf', class_weight='balanced', random_state=42), True),
]

for name, clf, use_small in tuned_svc:
    print(f"\n>>> {name}")
    trs = time()
    clf.fit(X_train_small, y_train_small)
    y_pred = clf.predict(X_test)
    tre = time() - trs
    
    all_results[name] = {
        'accuracy': accuracy_score(y_test, y_pred),
        'balanced_accuracy': balanced_accuracy_score(y_test, y_pred),
        'f1_score': f1_score(y_test, y_pred, average='weighted'),
        'mcc': matthews_corrcoef(y_test, y_pred),
        'time': tre,
        'predictions': y_pred
    }
    print(f"    Time: {tre:.1f}s | Acc: {all_results[name]['accuracy']:.4f} | Bal_Acc: {all_results[name]['balanced_accuracy']:.4f}")
```

```python
# ============================================================
# STEP 3: Tuned LinearSVC (Fast Alternative)
# ============================================================

tuned_linearsvc = [
    ("LinearSVC_balanced", LinearSVC(class_weight='balanced', random_state=42, max_iter=2000), False),
    ("LinearSVC_C10", LinearSVC(C=10, random_state=42, max_iter=2000), False),
    ("LinearSVC_C0.1_balanced", LinearSVC(C=0.1, class_weight='balanced', random_state=42, max_iter=2000), False),
]

for name, clf, use_small in tuned_linearsvc:
    print(f"\n>>> {name}")
    trs = time()
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    tre = time() - trs
    
    all_results[name] = {
        'accuracy': accuracy_score(y_test, y_pred),
        'balanced_accuracy': balanced_accuracy_score(y_test, y_pred),
        'f1_score': f1_score(y_test, y_pred, average='weighted'),
        'mcc': matthews_corrcoef(y_test, y_pred),
        'time': tre,
        'predictions': y_pred
    }
    print(f"    Time: {tre:.1f}s | Acc: {all_results[name]['accuracy']:.4f} | Bal_Acc: {all_results[name]['balanced_accuracy']:.4f}")
```

```python
# ============================================================
# STEP 4: Tuned Ridge Classifier
# ============================================================

tuned_ridge = [
    ("Ridge_balanced", RidgeClassifier(class_weight='balanced', random_state=42), False),
    ("Ridge_alpha0.1", RidgeClassifier(alpha=0.1, random_state=42), False),
]

for name, clf, use_small in tuned_ridge:
    print(f"\n>>> {name}")
    trs = time()
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    tre = time() - trs
    
    all_results[name] = {
        'accuracy': accuracy_score(y_test, y_pred),
        'balanced_accuracy': balanced_accuracy_score(y_test, y_pred),
        'f1_score': f1_score(y_test, y_pred, average='weighted'),
        'mcc': matthews_corrcoef(y_test, y_pred),
        'time': tre,
        'predictions': y_pred
    }
    print(f"    Time: {tre:.1f}s | Acc: {all_results[name]['accuracy']:.4f} | Bal_Acc: {all_results[name]['balanced_accuracy']:.4f}")
```

```python
# ============================================================
# RESULTS TABLE
# ============================================================

results_df = pandas.DataFrame({k: {m: v[m] for m in ['accuracy', 'balanced_accuracy', 'f1_score', 'mcc', 'time']} 
                               for k, v in all_results.items()}).T
results_df = results_df.sort_values('balanced_accuracy', ascending=False)

print("\n" + "="*70)
print("RESULTS - SORTED BY BALANCED ACCURACY")
print("="*70)
print("\n" + results_df.to_string())
```

```python
# ============================================================
# VISUALIZATION: Performance Comparison
# ============================================================

fig, ax = plt.subplots(figsize=(12, 7))

plot_df = results_df.sort_values('balanced_accuracy', ascending=True)
colors = ['#2ecc71' if x == plot_df['balanced_accuracy'].max() else '#3498db' for x in plot_df['balanced_accuracy']]

bars = ax.barh(plot_df.index, plot_df['balanced_accuracy'], color=colors, edgecolor='black')
ax.axvline(x=all_results['SVC_baseline']['balanced_accuracy'], color='red', linestyle='--', 
           linewidth=2, label=f"SVC Baseline ({all_results['SVC_baseline']['balanced_accuracy']:.4f})")

ax.set_xlabel('Balanced Accuracy', fontsize=12)
ax.set_title('SUPPORT VECTOR CLASSIFIER COMPARISON\n(Green = Best, Red Line = SVC Baseline)', 
             fontweight='bold', fontsize=14)
ax.set_xlim(0.75, 0.90)
ax.legend(loc='lower right')

for bar, val in zip(bars, plot_df['balanced_accuracy']):
    ax.annotate(f'{val:.4f}', xy=(val, bar.get_y() + bar.get_height()/2),
                xytext=(5, 0), textcoords='offset points', va='center', fontsize=9)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# FINAL CONCLUSION
# ============================================================

best_model = results_df['balanced_accuracy'].idxmax()
best_ba = results_df.loc[best_model, 'balanced_accuracy']
svc_baseline_ba = all_results['SVC_baseline']['balanced_accuracy']
improvement = ((best_ba - svc_baseline_ba) / svc_baseline_ba) * 100

print("="*70)
print("FINAL CONCLUSION")
print("="*70)

print(f"""
SUPPORT VECTOR CLASSIFIER ANALYSIS:

BASELINE PERFORMANCE:
  SVC Baseline:       {svc_baseline_ba:.4f}
  Best Tuned Model:   {best_ba:.4f} ({best_model})
  
  IMPROVEMENT: {improvement:+.2f}%

KEY INSIGHTS:
  1. SVC with RBF kernel provides best accuracy but is SLOW
  2. LinearSVC is much FASTER (trains on full dataset)
  3. class_weight='balanced' helps with imbalanced intrusion data
  4. Ridge Classifier is fastest but lower accuracy

BEST MODEL METRICS ({best_model}):
  - Accuracy:          {all_results[best_model]['accuracy']:.4f}
  - Balanced Accuracy: {all_results[best_model]['balanced_accuracy']:.4f}
  - F1-Score:          {all_results[best_model]['f1_score']:.4f}
  - MCC:               {all_results[best_model]['mcc']:.4f}
  - Training Time:     {all_results[best_model]['time']:.2f}s

RECOMMENDATION:
  For production use, consider LinearSVC with class_weight='balanced'
  as it trains on full data and is much faster than SVC.
""")

print("="*70)
```
